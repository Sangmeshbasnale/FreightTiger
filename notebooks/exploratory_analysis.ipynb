# FreightTiger: Shipping Cost Exploratory Data Analysis (EDA)

This notebook explores the shipment records and cost trends across corridors, validating weekly aggregations and baseline trends.

```python
import pandas as pd
import numpy as np
from pathlib import Path

# Load clean weekly features and submission
weekly_df = pd.read_csv('../outputs/route_week_features.csv')
sub_df = pd.read_csv('../outputs/submission.csv')
notes_df = pd.read_csv('../data/raw/context_notes.csv')

print(f"Total route-weeks: {len(weekly_df)}")
print(f"Total anomalies flagged: {(sub_df['flagged'] != 'No').sum()}")
sub_df.head()
```

### 1. Cost Efficiency Summary by Corridor

```python
summary = weekly_df.groupby(['route', 'route_type'])['cost_per_tonne_km'].agg(['min', 'mean', 'max', 'std']).reset_index()
summary.columns = ['Route', 'Length Bucket', 'Min INR', 'Average INR', 'Max INR', 'Std Dev']
summary
```

### 2. Anomaly Candidates and Matched Context Notes

```python
anomalies = sub_df[sub_df['flagged'] != 'No'][['route', 'week_of', 'cost_per_tonne_km', 'vs_own_history', 'vs_similar_routes', 'flagged', 'matched_note_id', 'reason']]
print(f"Justified: {(anomalies['flagged'] == 'No (justified)').sum()}")
print(f"Unexplained: {(anomalies['flagged'] == 'Yes').sum()}")
anomalies.head(10)
```
